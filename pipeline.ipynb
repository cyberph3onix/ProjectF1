```python
def build_race_features(year,gp):
    race = fastf1.get_session(year,gp,"R")
    quali = fastf1.get_session(year,gp,"Q")

    race.load()
    quali.load()

    laps = race.laps
    results = race.results

    avg_race_pace = laps.groupby("Driver")["LapTime"].mean()
    max_stint = laps.groupby("Driver")["Stint"].max()
    pitstops = laps.groupby("Driver")["PitInTime"].count()
    quali_pace = quali.laps.groupby("Driver")["LapTime"].min()
    grid_pos = results.set_index("Abbreviation")["GridPosition"]
    winner = (results.set_index("Abbreviation")["Position"] == 1).astype(int)
    df = pd.DataFrame({
        "avg_race_pace": avg_race_pace,
        "max_stint": max_stint,
        "pitstops": pitstops,
        "quali_pace": quali_pace,
        "grid_position": grid_pos,
        "win": winner
    }).reset_index()

    df["season"] = year
    df["gp"] = gp

    return df.dropna()
```

```python
def build_full_dataset(start=2022, end=2024):
    all_data = []

    for year in range(start, end + 1):
        try:
            # ✅ Force Ergast fallback if FastF1 backend fails
            schedule = fastf1.get_event_schedule(year, backend='ergast')
            print(f"📅 Loaded schedule for {year}")
        except Exception as e:
            print(f"❌ Failed to load schedule for {year} | {str(e)[:60]}")
            continue   # Skip this year safely

        for gp in schedule["EventName"]:
            try:
                df = build_race_features(year, gp)
                all_data.append(df)
                print(f"✅ {year} {gp}")
            except Exception as e:
                print(f"❌ Skipped {year} {gp} | {str(e)[:60]}")

    if len(all_data) == 0:
        raise RuntimeError("❌ No race data could be loaded at all.")

    return pd.concat(all_data, ignore_index=True)
```

```python
import os
import re
import time
import fastf1
import pandas as pd
from pathlib import Path
# --- helper to sanitize GP names for filenames ---
def safe_name(s: str) -> str:
    return re.sub(r'[^A-Za-z0-9_\-]+', '_', s).strip('_')

# --- your existing feature-extraction function ---
def build_race_features(year, gp):
    # same as your function; ensure it returns df for that race
    race = fastf1.get_session(year, gp, "R")
    quali = fastf1.get_session(year, gp, "Q")
    race.load()
    quali.load()

    laps = race.laps
    results = race.results

    avg_race_pace = laps.groupby("Driver")["LapTime"].mean()
    max_stint = laps.groupby("Driver")["Stint"].max()
    pitstops = laps.groupby("Driver")["PitInTime"].count()
    quali_pace = quali.laps.groupby("Driver")["LapTime"].min()
    grid_pos = results.set_index("Abbreviation")["GridPosition"]
    winner = (results.set_index("Abbreviation")["Position"] == 1).astype(int)

    df = pd.DataFrame({
        "avg_race_pace": avg_race_pace,
        "max_stint": max_stint,
        "pitstops": pitstops,
        "quali_pace": quali_pace,
        "grid_position": grid_pos,
        "win": winner
    }).reset_index().rename(columns={"index":"Driver"})

    df["season"] = year
    df["gp"] = gp
    return df.dropna()

# --- robust resumable dataset builder ---
def build_full_dataset_resumable(start=2023, end=2024, out_dir="data", max_gp_retries=5):
    os.makedirs(out_dir, exist_ok=True)
    saved_files = set([p.name for p in Path(out_dir).glob("*.csv")])

    all_files = []  # will store filenames we succeeded for final concat

    for year in range(start, end + 1):
        try:
            schedule = fastf1.get_event_schedule(year, force_ergast=True)
            print(f"📅 Loaded schedule for {year} ({len(schedule)} events)")
        except Exception as e:
            print(f"❌ Failed to load schedule for {year} | {e}")
            continue

        for gp in schedule["EventName"]:
            fname = f"{year}_{safe_name(gp)}.csv"
            fpath = os.path.join(out_dir, fname)

            if fname in saved_files:
                print(f"⏭️ Skipping (already saved): {fname}")
                all_files.append(fpath)
                continue

            # Try multiple times with exponential backoff
            succeeded = False
            for attempt in range(1, max_gp_retries + 1):
                try:
                    print(f"🔁 [{attempt}/{max_gp_retries}] Loading {year} - {gp} ...")
                    df_race = build_race_features(year, gp)

                    # Save immediate to csv so we never lose progress
                    df_race.to_csv(fpath, index=False)
                    print(f"✅ Saved: {fpath} ({len(df_race)} rows)")
                    all_files.append(fpath)
                    saved_files.add(fname)
                    succeeded = True
                    # short sleep to be polite
                    time.sleep(0.5)
                    break
                except Exception as e:
                    wait = min(60, 2 ** attempt)  # exponential backoff capped at 60s
                    print(f"⚠ Failed ({attempt}) for {year} {gp} | {str(e)[:120]} -> retry in {wait}s")
                    time.sleep(wait)
            if not succeeded:
                print(f"❌ GAVE UP on {year} {gp} after {max_gp_retries} attempts. Continue to next GP.")

    if len(all_files) == 0:
        raise RuntimeError("❌ No race CSVs were created. Check earlier logs or try a different network.")

    # concat all saved CSVs (only those that exist)
    df_list = []
    for p in all_files:
        if os.path.exists(p):
            try:
                df_list.append(pd.read_csv(p))
            except Exception as e:
                print(f"⚠ Failed to read {p}: {e}")

    df_all = pd.concat(df_list, ignore_index=True)
    print(f"🏁 Final dataset shape: {df_all.shape}")
    return df_all

# ---- run it ----
df = build_full_dataset_resumable(2023, 2024, out_dir="data", max_gp_retries=5)
```

```python
# =========================
# 0) IMPORTS
# =========================
import numpy as np
import pandas as pd
import os
from glob import glob
import joblib
import matplotlib.pyplot as plt

from sklearn.metrics import log_loss, accuracy_score, brier_score_loss
from sklearn.model_selection import train_test_split
from lightgbm import LGBMClassifier
from lightgbm import early_stopping, log_evaluation


# =========================
# 1) LOAD DATASET FROM SAVED CSVs
# =========================
csv_files = glob("data/*.csv")

print("Found CSV files:", len(csv_files))
if len(csv_files) == 0:
    raise RuntimeError("❌ No CSV files found in data/ folder.")

df = pd.concat([pd.read_csv(f) for f in csv_files], ignore_index=True)
print("✅ Restored df shape:", df.shape)
display(df.head())


# =========================
# 2) FIX BAD DTYPES (object -> float)
# =========================
for col in ['avg_race_pace', 'quali_pace']:
    if col in df.columns:
        if df[col].dtype == 'object':
            df[col] = pd.to_timedelta(df[col], errors='coerce').dt.total_seconds()

df['avg_race_pace'] = pd.to_numeric(df['avg_race_pace'], errors='coerce')
df['quali_pace'] = pd.to_numeric(df['quali_pace'], errors='coerce')

df = df.dropna(subset=['avg_race_pace', 'quali_pace'])

print("\n✅ dtypes after fix:")
print(df[['avg_race_pace','quali_pace']].dtypes)
print("✅ df shape after cleaning:", df.shape)


# =========================
# 3) FEATURE PREP
# =========================
features = ['avg_race_pace','max_stint','pitstops','quali_pace','grid_position']

X = df[features].copy()
y = df['win'].astype(int)
meta = df[['season','gp','Driver']].copy()


# =========================
# 4) TIME-AWARE SPLIT
# =========================
train_mask = df['season'] <= 2022
val_mask   = df['season'] == 2023
test_mask  = df['season'] == 2024

X_train, y_train = X[train_mask], y[train_mask]
X_val,   y_val   = X[val_mask],   y[val_mask]
X_test,  y_test  = X[test_mask],  y[test_mask]

print("\nShapes -> train:", X_train.shape, "val:", X_val.shape, "test:", X_test.shape)

# fallback if season split is empty
if X_train.shape[0] == 0 or X_test.shape[0] == 0:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, shuffle=False
    )
    print("⚠️ Fell back to chronological split:", X_train.shape, X_test.shape)


# =========================
# 5) TRAIN LIGHTGBM (MODERN API)
# =========================
model = LGBMClassifier(
    objective='binary',
    learning_rate=0.05,
    n_estimators=1000,
    num_leaves=31,
    random_state=42
)

model.fit(
    X_train, y_train,
    eval_set=[(X_val, y_val)],
    eval_metric='binary_logloss',
    callbacks=[
        early_stopping(stopping_rounds=50),
        log_evaluation(period=50)
    ]
)


# =========================
# 6) PREDICTIONS
# =========================
y_prob = model.predict_proba(X_test)[:,1]


# =========================
# 7) GLOBAL METRICS
# =========================
ll = log_loss(y_test, y_prob)
acc = accuracy_score(y_test, (y_prob > 0.5).astype(int))
brier = brier_score_loss(y_test, y_prob)

print("\n===== GLOBAL TEST METRICS =====")
print(f"Log Loss: {ll:.4f}")
print(f"Accuracy (0.5 cutoff): {acc:.4f}")
print(f"Brier Score: {brier:.4f}")


# =========================
# 8) PER-RACE TOP-K ACCURACY
# =========================
meta_test = meta[test_mask].reset_index(drop=True)
X_test_reset = X_test.reset_index(drop=True)
y_test_reset = y_test.reset_index(drop=True)

pred_df = pd.concat([meta_test, X_test_reset], axis=1)
pred_df['pred_prob'] = y_prob
pred_df['true_win'] = y_test_reset.values


def race_topk_accuracy(pred_df, k=1):
    scores = []
    for (s, gp), grp in pred_df.groupby(['season','gp']):
        grp_sorted = grp.sort_values('pred_prob', ascending=False)
        topk = grp_sorted.head(k)['Driver'].values
        winner_present = grp.loc[grp['true_win'] == 1, 'Driver'].isin(topk).any()
        scores.append(1 if winner_present else 0)
    return np.mean(scores)


top1 = race_topk_accuracy(pred_df, k=1)
top3 = race_topk_accuracy(pred_df, k=3)

print("\n===== PER-RACE METRICS =====")
print(f"Top-1 Accuracy: {top1:.4f}")
print(f"Top-3 Accuracy: {top3:.4f}")


# =========================
# 9) FEATURE IMPORTANCES
# =========================
fi = pd.DataFrame({
    'feature': features,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

print("\n===== FEATURE IMPORTANCES =====")
print(fi.to_string(index=False))

plt.figure(figsize=(6,3))
plt.bar(fi['feature'], fi['importance'])
plt.title('Feature Importances (LightGBM)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# =========================
# 10) SAVE MODEL
# =========================
joblib.dump(model, "model.joblib")
print("\n✅ Model saved as model.joblib")
```
